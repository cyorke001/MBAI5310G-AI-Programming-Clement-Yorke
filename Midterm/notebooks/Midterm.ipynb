## Business Context

TalentWorks is an HR consultancy helping mid-size organisations reduce employee attrition. This notebook applies **K-Means clustering** to identify natural groupings of employees based on their workplace characteristics. Clustering is unsupervised: we do not use the `Left_Company` column to train the model. It is used only after clustering to validate whether the segments are meaningful for attrition risk.
